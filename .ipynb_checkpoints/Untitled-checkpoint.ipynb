```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# 1. Forçamos o Plotly a renderizar corretamente no Jupyter
pio.renderers.default = 'iframe' # Se ainda ficar em branco, troque 'iframe' por 'notebook' ou 'browser'

# 2. Criamos uns dados simples simulados
dados = pd.DataFrame({
    'Produto': ['Bicicletas', 'Acessórios', 'Roupas'],
    'Vendas': [150, 80, 45]
})

# 3. Geramos o gráfico interativo
fig = px.bar(dados, x='Produto', y='Vendas', title='O Nosso Primeiro Gráfico Interativo')

# 4. Mostramos o gráfico no ecrã
fig.show()
```

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
from io import StringIO

csv_data = """Categoria,Territorio,Vendas,Ano
Bikes,North America,151500000,2025
Bikes,Europe,90000,2025
Acessories,North America,300000,2025
Acessories,Europe,150000,2025
Clothing,North America,200000,2025
Clothing,Europe,100000,2025
Bikes,North America,1800000,2026
Bikes,Europe,1100000,2026
Acessories,North America,400000,2026
Acessories,Europe,200000,2026
Clothing,North America,250000,2026
Clothing,Europe,120000,2026
"""
df = pd.read_csv(StringIO(csv_data))

app = dash.Dash(__name__)

app.layout = html.Div(style={'fontFamily': 'Segoe UI, sans-serif', 'padding': '20px'}, children=[
    html.H2("AdventureWorks - Análise de Vendas", style={'color': '#333'}),
    html.P("Demonstração AP2: Projeto de Desenvolvimento de Dashboard Analítico", style={'color': '#666'}),

    html.Div([
        html.Label("Selecione o Ano Fiscal:"),
        dcc.Dropdown(
            id='filtro-ano',
            options=[{'label': str(ano), 'value': ano} for ano in df['Ano'].unique()],
            value=2026,
            clearable=False,
            style={'width': '300px', 'marginTop': '5px'}
        )
   ], style={'marginBottom': '30px'}),

   dcc.Graph(id='grafico-vendas')
])       

@app.callback(
    Output('grafico-vendas', 'figure'),
    Input('filtro-ano', 'value')
)
def atualizar_dashboard(ano_selecionado):
    df_filtrado = df[df['Ano'] == ano_selecionado]

    fig = px.bar(
        df_filtrado,
        x='Categoria',
        y='Vendas',
        color='Territorio',
        barmode='group',
        title=f'Faturamento por Categoria e Territorio ({ano_selecionado})',
        text_auto='.2s'
    )

    fig.update_layout(
        template='plotly_white',
        yaxis_title="Faturamento Bruto ($)",
        xaxis_title="Categoria de Produto"
    )

    return fig

if __name__ == '__main__':
    app.run(jupyter_mode='inline', port=8050)
```
