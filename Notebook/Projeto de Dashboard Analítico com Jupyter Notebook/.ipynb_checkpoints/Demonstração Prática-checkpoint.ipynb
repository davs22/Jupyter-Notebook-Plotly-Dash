```python
import pandas as pd
from sqlalchemy import create_engine
import plotly.express as px
from dash import Dash, dsn, html, dcc, Input, Output
import dash_bootstrap_components as dbc

# ==========================================
# 1. ETAPA ETL (Extração, Transformação, Carga)
# ==========================================

# Substitua pelo nome do seu servidor SQL Server
server = 'LOCALHOST\\SQLEXPRESS' 
driver = 'ODBC+Driver+17+for+SQL+Server'

# Criar conexões com os bancos de dados
engine_oltp = create_engine(f"mssql+pyodbc://@{server}/AdventureWorks2022?driver={driver}&trusted_connection=yes")
engine_dw = create_engine(f"mssql+pyodbc://@{server}/AdventureWorksDW2022?driver={driver}&trusted_connection=yes")

print("🔄 Iniciando processo de ETL...")

# --- EXTRAÇÃO & TRANSFORMAÇÃO (Exemplo: Clientes) ---
query_clientes = """
SELECT 
    c.CustomerID AS CustomerAlternateKey,
    p.FirstName + ' ' + p.LastName AS FullName,
    p.EmailPromotion,
    CASE WHEN a.City IS NULL THEN 'Unknown' ELSE a.City END AS City
FROM Sales.Customer c
INNER JOIN Person.Person p ON c.PersonID = p.BusinessEntityID
LEFT JOIN Person.BusinessEntityAddress bea ON p.BusinessEntityID = bea.BusinessEntityID
LEFT JOIN Person.Address a ON bea.AddressID = a.AddressID
"""
df_clientes = pd.read_sql(query_clientes, engine_oltp)

# --- CARGA (Inserindo no Data Warehouse) ---
# Usamos if_exists='append' para simular uma carga incremental ou 'replace' para testes
df_clientes.to_sql('DimCustomer_ETL_Stage', con=engine_dw, if_exists='replace', index=False)
print("✅ Dados de clientes migrados para o DW!")

# --- EXTRAÇÃO PARA O DASHBOARD (Lendo dados agregados do DW) ---
# Vamos puxar os dados da FactInternetSales original do DW para criar os gráficos
query_vendas = """
SELECT 
    f.SalesAmount, 
    f.OrderDate,
    g.EnglishCountryRegionName AS Country,
    p.EnglishProductName AS Product
FROM FactInternetSales f
JOIN DimCustomer c ON f.CustomerKey = c.CustomerKey
JOIN DimGeography g ON c.GeographyKey = g.GeographyKey
JOIN DimProduct p ON f.ProductKey = p.ProductKey
"""
df_vendas = pd.read_sql(query_vendas, engine_dw)
df_vendas['OrderDate'] = pd.to_datetime(df_vendas['OrderDate'])
df_vendas['Year'] = df_vendas['OrderDate'].dt.year

print("📊 ETL concluído. Inicializando o Dashboard no Dash...")

# ==========================================
# 2. CONSTRUÇÃO DO DASHBOARD (Dash + Plotly)
# ==========================================

# Inicializar o app Dash específico para rodar dentro do Jupyter (utiliza o servidor embutido)
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

# Layout do Dashboard
app.layout = dbc.Container([
    dbc.Row([
        dbc.Col(html.H1("Dashboard de Vendas - AdventureWorks DW", className="text-center my-4 text-primary"), width=12)
    ]),
    
    dbc.Row([
        dbc.Col([
            html.Label("Selecione o País:"),
            dcc.Dropdown(
                id='dropdown-pais',
                options=[{'label': c, 'value': c} for c in df_vendas['Country'].unique()],
                value=df_vendas['Country'].unique()[0],
                clearable=False
            )
        ], width=4),
        dbc.Col([
            html.Label("Selecione o Ano:"),
            dcc.Dropdown(
                id='dropdown-ano',
                options=[{'label': str(y), 'value': y} for y in sorted(df_vendas['Year'].unique())],
                value=sorted(df_vendas['Year'].unique())[-1],
                clearable=False
            )
        ], width=4)
    ], className="mb-4"),
    
    dbc.Row([
        dbc.Col(dcc.Graph(id='grafico-faturamento-produtos'), width=6),
        dbc.Col(dcc.Graph(id='grafico-evolucao-vendas'), width=6)
    ])
], fluid=True)

# Callbacks para atualizar os gráficos interativamente
@app.callback(
    [Output('grafico-faturamento-produtos', 'figure'),
     Output('grafico-evolucao-vendas', 'figure')],
    [Input('dropdown-pais', 'value'),
     Input('dropdown-ano', 'value')]
)
def atualizar_graficos(pais_selecionado, ano_selecionado):
    # Filtrar o DataFrame
    df_filtrado = df_vendas[(df_vendas['Country'] == pais_selecionado) & (df_vendas['Year'] == ano_selecionado)]
    
    # 1. Gráfico de Barras: Top Produtos por Faturamento
    df_produtos = df_filtrado.groupby('Product')['SalesAmount'].sum().reset_index()
    df_produtos = df_produtos.sort_values(by='SalesAmount', ascending=False).head(10)
    
    fig_produtos = px.bar(
        df_produtos, x='SalesAmount', y='Product', orientation='h',
        title=f"Top 10 Produtos em {pais_selecionado} ({ano_selecionado})",
        labels={'SalesAmount': 'Faturamento ($)', 'Product': 'Produto'},
        template='plotly_white'
    )
    fig_produtos.update_layout(yaxis={'categoryorder':'total ascending'})
    
    # 2. Gráfico de Linha: Evolução de Vendas Mensal
    df_mensal = df_filtrado.copy()
    df_mensal['Month'] = df_mensal['OrderDate'].dt.to_period('M').astype(str)
    df_linha = df_mensal.groupby('Month')['SalesAmount'].sum().reset_index()
    
    fig_linha = px.line(
        df_linha, x='Month', y='SalesAmount',
        title=f"Evolução Mensal do Faturamento em {ano_selecionado}",
        labels={'SalesAmount': 'Faturamento ($)', 'Month': 'Mês'},
        template='plotly_white', markers=True
    )
    
    return fig_produtos, fig_linha

# Rodar o servidor inline no Jupyter Notebook
if __name__ == '__main__':
    app.run_server(mode='inline', port=8050)
```
