```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
from io import StringIO

# 1. Base de Dados: Simulação do AdventureWorks (Vendas por Território e Categoria)
csv_data = """Categoria,Territorio,Vendas,Ano
Bikes,North America,1500000,2025
Bikes,Europe,900000,2025
Accessories,North America,300000,2025
Accessories,Europe,150000,2025
Clothing,North America,200000,2025
Clothing,Europe,100000,2025
Bikes,North America,1800000,2026
Bikes,Europe,1100000,2026
Accessories,North America,400000,2026
Accessories,Europe,200000,2026
Clothing,North America,250000,2026
Clothing,Europe,120000,2026
"""
df = pd.read_csv(StringIO(csv_data))

# 2. Inicialização do App Dash (compatível com Jupyter)
app = dash.Dash(__name__)

# 3. Layout: Construção da Interface Web (Sem usar HTML/CSS externo)
app.layout = html.Div(style={'fontFamily': 'Segoe UI, sans-serif', 'padding': '20px'}, children=[
    html.H2("AdventureWorks - Análise de Vendas", style={'color': '#333'}),
    html.P("Demonstração AP2: Projeto de Desenvolvimento de Dashboard Analítico", style={'color': '#666'}),
    
    # Componente Interativo: Filtro de Ano
    html.Div([
        html.Label("Selecione o Ano Fiscal:"),
        dcc.Dropdown(
            id='filtro-ano',
            options=[{'label': str(ano), 'value': ano} for ano in df['Ano'].unique()],
            value=2026,
            clearable=False,
            style={'width': '300px', 'marginTop': '5px'}
        )
    ], style={'marginBottom': '30px'}),
    
    # Componente Visual: Gráfico Plotly
    dcc.Graph(id='grafico-vendas')
])

# 4. Callbacks: Onde a interatividade acontece
@app.callback(
    Output('grafico-vendas', 'figure'),
    Input('filtro-ano', 'value')
)
def atualizar_dashboard(ano_selecionado):
    # Lógica de Filtro (Pandas)
    df_filtrado = df[df['Ano'] == ano_selecionado]
    
    # Geração do Gráfico Interativo (Plotly)
    fig = px.bar(
        df_filtrado, 
        x='Categoria', 
        y='Vendas', 
        color='Territorio',
        barmode='group',
        title=f'Faturamento por Categoria e Território ({ano_selecionado})',
        text_auto='.2s'
    )
    
    # Estilização do Gráfico
    fig.update_layout(
        template='plotly_white',
        yaxis_title="Faturamento Bruto ($)",
        xaxis_title="Categoria de Produto"
    )
    
    return fig

# 5. Execução do Servidor no Jupyter
if __name__ == '__main__':
    # CORREÇÃO: app.run no lugar de app.run_server
    app.run(jupyter_mode='inline', port=8050)
```
