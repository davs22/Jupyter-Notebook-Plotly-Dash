```python
import pandas as pd

# 1. Caminho real do arquivo CSV do AdventureWorks na sua máquina
caminho_arquivo = 'C:/Users/mdiir/Documents/code_unisales/Database_Tunning/AdventureWorks-for-Postgres-master/SalesOrderDetail.csv'

# 2. Definindo os nomes reais das colunas da tabela SalesOrderDetail do AdventureWorks
colunas_sales_order_detail = [
    'SalesOrderID', 'SalesOrderDetailID', 'CarrierTrackingNumber', 
    'OrderQty', 'ProductID', 'SpecialOfferID', 'UnitPrice', 
    'UnitPriceDiscount', 'LineTotal', 'rowguid', 'ModifiedDate'
]

# 3. EXTRAÇÃO (Extract) em memória usando Pandas (separador por tabulação '\t', sem cabeçalho original)
print("=== ETAPA 1: EXTRAÇÃO EM MEMÓRIA ===")
df_real = pd.read_csv(caminho_arquivo, sep='\t', header=None, names=colunas_sales_order_detail)
print(f"Total de registros carregados na RAM: {len(df_real)}")

# 4. TRANSFORMAÇÃO (Transform) em memória
print("\n=== ETAPA 2: TRANSFORMAÇÃO EM MEMÓRIA ===")
# Removendo duplicatas, se houver
df_transformado = df_real.drop_duplicates().copy()

# Tratamento de valores nulos nas colunas numéricas
df_transformado['OrderQty'] = df_transformado['OrderQty'].fillna(0)
df_transformado['UnitPrice'] = df_transformado['UnitPrice'].fillna(0.0)

# Criando métrica derivada em memória (Regra de Negócio: Faturamento Total da Linha com Desconto)
df_transformado['LineTotal_Calculated'] = (
    df_transformado['OrderQty'] * df_transformado['UnitPrice']
) * (1 - df_transformado['UnitPriceDiscount'])

print(f"Total de registros após limpeza e transformação: {len(df_transformado)}")

# 5. Exibindo uma amostra dos dados processados em memória
print("\n=== AMOSTRA DOS DADOS PROCESSADOS (HEAD) ===")
display(df_transformado[['SalesOrderID', 'OrderQty', 'UnitPrice', 'UnitPriceDiscount', 'LineTotal_Calculated']].head(10))

# 6. Agregação analítica simples em memória (exemplo de KPI: Faturamento Total por Pedido)
df_resumo_pedido = df_transformado.groupby('SalesOrderID')['LineTotal_Calculated'].sum().reset_index()
df_resumo_pedido.rename(columns={'LineTotal_Calculated': 'TotalFaturamento'}, inplace=True)

print("\n=== RESUMO ANALÍTICO EM MEMÓRIA (TOP 5 PEDIDOS) ===")
display(df_resumo_pedido.sort_values(by='TotalFaturamento', ascending=False).head(5))
```
